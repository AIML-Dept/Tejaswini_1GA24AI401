```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

sim = AerSimulator()

print("Ready")
```

```python
#Easy: Construct the Phi+ Bell state and confirm through 1024-shot simulation that only '00' and '11' outcomes occur.
qc = QuantumCircuit(2, 2)

qc.h(0)
qc.cx(0, 1)

qc.measure([0, 1], [0, 1])

result = sim.run(qc, shots=1024).result()
print(result.get_counts())
```

```python
#Medium: Construct all four Bell states and tabulate the measurement outcome distributions for comparison. 
states = ["Phi+", "Phi-", "Psi+", "Psi-"]

for state in states:
    qc = QuantumCircuit(2, 2)

    if state == "Phi+":
        qc.h(0)
        qc.cx(0, 1)

    elif state == "Phi-":
        qc.h(0)
        qc.cx(0, 1)
        qc.z(0)

    elif state == "Psi+":
        qc.h(0)
        qc.cx(0, 1)
        qc.x(0)

    elif state == "Psi-":
        qc.h(0)
        qc.cx(0, 1)
        qc.x(0)
        qc.z(0)

    qc.measure([0, 1], [0, 1])

    result = sim.run(qc, shots=1024).result()
    print(state, ":", result.get_counts())
```

```python
# Exercise 3: Measure a Bell pair in the Hadamard basis and observe the correlation pattern.

qc = QuantumCircuit(2, 2)

# Create Phi+ Bell state
qc.h(0)
qc.cx(0, 1)

# Measure in Hadamard basis
qc.h(0)
qc.h(1)

qc.measure([0, 1], [0, 1])

result = sim.run(qc, shots=1024).result()
print("Hadamard Basis:", result.get_counts())
```

```python
# Exercise 4: Relate Bell-state correlations conceptually to Quantum Key Distribution (QKD).

qc = QuantumCircuit(2, 2)

# Create entangled Bell pair
qc.h(0)
qc.cx(0, 1)

qc.measure([0, 1], [0, 1])

result = sim.run(qc, shots=1024).result()
print("QKD Bell Pair:", result.get_counts())
```

```python
# Exercise 5: Construct a 3-qubit GHZ state and examine its entanglement structure.

qc = QuantumCircuit(3, 3)

qc.h(0)
qc.cx(0, 1)
qc.cx(0, 2)

qc.measure([0, 1, 2], [0, 1, 2])

result = sim.run(qc, shots=1024).result()
print("GHZ State:", result.get_counts())
```

```python
# Exercise 5: Construct a 3-qubit W state and compare its entanglement structure with GHZ.

qc = QuantumCircuit(3, 3)

# Prepare W state
qc.ry(2 * 1.230959, 0)
qc.cx(0, 1)
qc.cry(1.910633, 0, 1)
qc.cx(1, 2)
qc.cry(1.910633, 1, 2)
qc.cx(0, 1)
qc.cry(-1.910633, 1, 2)

qc.measure([0, 1, 2], [0, 1, 2])

result = sim.run(qc, shots=1024).result()
print("W State:", result.get_counts())
```
