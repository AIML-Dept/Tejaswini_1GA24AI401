```python
#4. Real-World – Estimate Rotation Angle
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import numpy as np

theta = 0.8

qc = QuantumCircuit(1, 1)
qc.ry(theta, 0)
qc.measure(0, 0)

result = AerSimulator().run(qc, shots=1000).result()
counts = result.get_counts()

p1 = counts.get('1', 0) / 1000

estimated = 2 * np.arcsin(np.sqrt(p1))

print("Measured probability:", p1)
print("Estimated angle:", estimated)
```

```python
#5. Challenge: Quantum Randomness vs Biased Classical Randomness
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from scipy.stats import chisquare
import numpy as np

shots = 1000

qc = QuantumCircuit(1, 1)
qc.h(0)
qc.measure(0, 0)

result = AerSimulator().run(qc, shots=shots).result()
quantum = result.get_counts()

print("Quantum:", quantum)

rng = np.random.default_rng(10)
bits = rng.choice([0, 1], shots, p=[0.3, 0.7])

classical = {
    '0': np.sum(bits == 0),
    '1': np.sum(bits == 1)
}

print("Classical:", classical)

qtest = chisquare(
    [quantum.get('0', 0), quantum.get('1', 0)],
    f_exp=[500, 500]
)

ctest = chisquare(
    [classical['0'], classical['1']],
    f_exp=[500, 500]
)

print("Quantum p-value:", qtest.pvalue)
print("Classical p-value:", ctest.pvalue)
```
