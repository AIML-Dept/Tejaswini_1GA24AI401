```python
# 1. Implement X, Y, and Z gates individually give gate 
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc_x = QuantumCircuit(1)
qc_x.x(0)

print("X Gate Circuit:")
print(qc_x.draw())

state_x = Statevector.from_instruction(qc_x)
print("State after X Gate:")
print(state_x)


qc_y = QuantumCircuit(1)
qc_y.y(0)

print("\nY Gate Circuit:")
print(qc_y.draw())

state_y = Statevector.from_instruction(qc_y)
print("State after Y Gate:")
print(state_y)

qc_z = QuantumCircuit(1)
qc_z.z(0)

print("\nZ Gate Circuit:")
print(qc_z.draw())

state_z = Statevector.from_instruction(qc_z)
print("State after Z Gate:")
print(state_z)
```

```python
#Construct the |−⟩ state using H and Z
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(1)

qc.h(0)

qc.z(0)

print("Quantum Circuit:")
print(qc.draw())

state = Statevector.from_instruction(qc)

print("\nFinal Statevector:")
print(state)
```

```python
#3 Apply five single-qubit gates in sequence (H, X, Z, Y, H) and observe the final quantum state using the Qiskit simulator.
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(1)

qc.h(0)   
qc.x(0)  
qc.z(0) 
qc.y(0)  
qc.h(0)   

print("Quantum Circuit:")
print(qc.draw())

state = Statevector.from_instruction(qc)

print("\nFinal Statevector:")
print(state)
```
