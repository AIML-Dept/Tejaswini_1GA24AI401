```python
#4 To simulate a Quantum Coin Flip using the Hadamard (H) gate and verify the measurement results using the Qiskit simulator.
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import matplotlib.pyplot as plt

qc = QuantumCircuit(1, 1)

qc.h(0)

qc.measure(0, 0)

print("Quantum Circuit:")
print(qc.draw())

simulator = AerSimulator()
job = simulator.run(qc, shots=1000)

result = job.result()
counts = result.get_counts()

print("\nMeasurement Results:")
print(counts)

plt.bar(counts.keys(), counts.values())
plt.title("Quantum Coin Flip Results")
plt.xlabel("Measurement Outcome")
plt.ylabel("Counts")
plt.show()
```

```python
# 5 ,To prove that applying the Hadamard (H) gate, followed by the Pauli-Z (Z) gate, and then another Hadamard (H) gate is equivalent to applying a single Pauli-X (X) gate.
from qiskit import QuantumCircuit
from qiskit.quantum_info import Operator

qc1 = QuantumCircuit(1)
qc1.h(0)
qc1.z(0)
qc1.h(0)

print("HZH Circuit:")
print(qc1.draw())

qc2 = QuantumCircuit(1)
qc2.x(0)

print("X Gate Circuit:")
print(qc2.draw())

print("\nMatrix of HZH:")
print(Operator(qc1).data)

print("\nMatrix of X Gate:")
print(Operator(qc2).data)
```
