```python
#To implement a classical full-adder circuit using CNOT and Toffoli gates.
from qiskit import QuantumCircuit

# 5 qubits:
# q0 = A
# q1 = B
# q2 = Cin
# q3 = Sum
# q4 = Cout

qc = QuantumCircuit(5)

qc.cx(0, 3)
qc.cx(1, 3)

qc.ccx(0, 1, 4)

qc.ccx(2, 3, 4)

qc.cx(2, 3)

print("Full Adder Circuit:")
print(qc.draw())
```

```python
#To create an entangled two-qubit state that is different from the standard Bell states and verify it using statevector inspection.
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
import numpy as np

theta = np.pi / 6

qc = QuantumCircuit(2)

qc.ry(2 * theta, 0)
qc.cx(0, 1)

print("Quantum Circuit:")
print(qc.draw())

state = Statevector.from_instruction(qc)

print("\nStatevector:")
print(state)

print("\nProbabilities:")
print(state.probabilities_dict())
```
